```python
import pandas as pd

PATH = "COVID-19-master\csse_covid_19_data\csse_covid_19_daily_reports/"

doc = pd.read_csv(f"{PATH}01-22-2020.csv", encoding="utf-8-sig")
```

```python
doc.head()
```

```python
doc = doc[["Country/Region", "Confirmed"]]
```

```python
doc
```

```python
doc.info()
```

```python
# 데이터 타입을 변경하고자 할 때! 우선적으로 결측치 값을 제거하고 시작해야 된다! (결측치는 타입 변경 시 에러를 발생시키기 때문이다)

doc = doc.dropna(subset=["Confirmed"])
doc.info()
```

```python
doc = doc.astype({"Confirmed": "int64"}) # 특정 컬럼의 타입 변경 시, 사용하는 메서드
doc.info()
```

```python
PATH = "COVID-19-master\csse_covid_19_data/"

doc = pd.read_csv(f"{PATH}UID_ISO_FIPS_LookUp_Table.csv", encoding="utf-8-sig")
```

```python
doc = doc[["iso2", "Country_Region"]]
doc
```

```python
# 중복되고 있는 값을 제거
# 2가지 선택지 : 첫 번째 최초로 발견된 값을 살리고, 나머지를 제거
# 마지막으로 사용된 값만 살리고, 그 앞에서 사용된 값들을 제거

doc = doc.drop_duplicates(subset="Country_Region", keep="last") # 만약, keep 사용하지 않는다면, first가 기본
doc # 4080개 행 중복 값 삭제하고 나서 202개의 행으로 줄어 들었다!
```
