```python
doc.describe() # 좀 더 세부적인 갓들을 보고 싶어!
# 전체 데이터를 중요한 지표만 추출해서 묘사한 것! 
# count : 행 수 / mean : 평균값 / std : 표준편차(평균으로부터 얼마나 떨어져 있는가에 대한 평균값) / min : 최소값 / max : 최대값
# 5 number summary
# FIPS는 왜 유독 없을까? FIPS = Federal Information Processing Standards Code = 미국 연방 지역 식별 코드 = 미국의 각각의 주, 도시의 식별값
# FIPS => Ex. 36061 : 뉴욕시 맨해튼 / 06037 : 미국 로스앤젤레스

# 데이터 수집해서 분석을 할 때 --> 높은 확률로 상관관계를 따지기 원해! 연결고리를 보려고 하는 것!
# 현재 우리의 데이터가 상관관계를 따져볼 수 있는 포맷(타입)으로 구성되어 있는가? --> 수학 공식을 통해 상관관계를 따짐! --> 모든 값들이 수치화된, 정형화된 값이 되어야 함!
# 보통은 이것이 쉽지 않아. 보려고 하는 것이 정형화된 데이터이기 쉽지 않아!
```

```python
# doc.corr() ---> doc에 문자열이 들어있기 때문에 error 가 날 거야!
# 문자열이 아닌 수치화된 데이터만 가져와서 상관관계를 따져보고 싶은 거야!
# 그럴 때 인자값으로 숫자만 갖고 오게 해야 됨!

doc.corr(numeric_only=True)
# 코로나 확진 판정 받으면 죽을 위험이 0.837153 : 상관관계가 높은 편
# 사망 확률도 높았지만, 회복할 수 있는 확률도 높았기 때문에 그래도 희망이 있다! 관리만 잘하면 회복이 가능한 병
# 확진 판정 받았으나 증상이 없는 사람도 있었다 물론 그 수는 매우 작았지만
# 이런 식으로 대략 판단! 디테일하게 판단 하려면 세부적으로 가야겠지!
```
